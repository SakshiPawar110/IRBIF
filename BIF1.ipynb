```python
# DNA Sequence Analysis
dna = "ATGCGCGAAATGCCCTAA"

# 1. Convert sequence to uppercase
dna = dna.upper()
print("DNA Sequence:", dna)

# 2. Calculate sequence length
print("Length:", len(dna))

# 3. Find a motif
motif = "ATG"
print("Motif Position:", dna.find(motif))

# 4. Calculate GC content
gc_count = dna.count("G") + dna.count("C")
gc_content = (gc_count / len(dna)) * 100
print("GC Content:", round(gc_content, 2), "%")

# 5. Find possible coding regions
start = dna.find("ATG")
stop_codons = ["TAA", "TAG", "TGA"]

if start != -1:
    for i in range(start + 3, len(dna) - 2, 3):
        codon = dna[i:i+3]
        if codon in stop_codons:
            print("Possible coding region:", dna[start:i+3])
            break
else:
    print("Start codon not found")
```
