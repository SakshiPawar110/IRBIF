````python
import glob
import subprocess

protein = "protein.pdbqt"

ligands = glob.glob("ligands/*.pdbqt")

for ligand in ligands:
    name = ligand.split("/")[-1].replace(".pdbqt", "")

    output = name + "_docked.pdbqt"
    log = name + "_log.txt"

    command = [
        "vina",
        "--receptor", protein,
        "--ligand", ligand,
        "--center_x", "10",
        "--center_y", "15",
        "--center_z", "20",
        "--size_x", "20",
        "--size_y", "20",
        "--size_z", "20",
        "--out", output,
        "--log", log
    ]

subprocess.run(command, check=True)
print("Docking completed:", name)
````

````python
import glob
import subprocess
import os

protein = "protein.pdbqt"

# Find all ligand files
ligands = glob.glob("ligands/*.pdbqt")

if not os.path.exists(protein):
    raise FileNotFoundError("protein.pdbqt not found!")

if not ligands:
    raise FileNotFoundError("No ligand PDBQT files found in ligands/")

# Dock each ligand
for ligand in ligands:
    name = os.path.splitext(os.path.basename(ligand))[0]

    output = name + "_docked.pdbqt"
    log = name + "_log.txt"

    command = [
        "vina",
        "--receptor", protein,
        "--ligand", ligand,
        "--center_x", "10",
        "--center_y", "15",
        "--center_z", "20",
        "--size_x", "20",
        "--size_y", "20",
        "--size_z", "20",
        "--out", output,
        "--log", log
    ]

    try:
        subprocess.run(command, check=True)
        print("Docking completed:", name)
    except subprocess.CalledProcessError as e:
        print("Docking failed:", name, "-", e)

print("All ligand docking attempts finished.")
````
