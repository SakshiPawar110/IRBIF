```python
documents = {
    1: "python is easy",
    2: "python is powerful",
    3: "java is powerful"
}

# Create inverted index
index = {}

for doc_id, text in documents.items():
    for word in text.split():
        index.setdefault(word, []).append(doc_id)

# Search a word
query = input("Enter search word: ").lower()

print("Inverted Index:", index)
print("Retrieved Documents:", index.get(query, []))
```
