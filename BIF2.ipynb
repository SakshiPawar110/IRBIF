```python
# RNA-Seq Data Analysis

import pandas as pd
import numpy as np

# Sample gene expression counts
data = {
    "Gene": ["Gene1", "Gene2", "Gene3", "Gene4"],
    "Control1": [100, 200, 50, 80],
    "Control2": [110, 190, 60, 90],
    "Treatment1": [200, 100, 40, 160],
    "Treatment2": [220, 110, 50, 180]
}

df = pd.DataFrame(data)
print("Original Data:")
print(df)

# Calculate average expression
df["Control_Mean"] = df[["Control1", "Control2"]].mean(axis=1)
df["Treatment_Mean"] = df[["Treatment1", "Treatment2"]].mean(axis=1)

# Calculate fold change
df["Fold_Change"] = (
    (df["Treatment_Mean"] + 1) /
    (df["Control_Mean"] + 1)
)

# Calculate log2 fold change
df["Log2FC"] = np.log2(df["Fold_Change"])

# Classify genes using a simple threshold
def classify(value):
    if value >= 1:
        return "Upregulated"
    elif value <= -1:
        return "Downregulated"
    else:
        return "No large change"

df["Result"] = df["Log2FC"].apply(classify)

print("\nDifferential Expression Results:")
print(df[["Gene", "Log2FC", "Result"]].round(2))
```
